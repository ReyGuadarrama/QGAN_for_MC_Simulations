```python
import torch
import matplotlib.pyplot as plt
```

```python
g1 = torch.ones(32).unsqueeze(0)
g2 = torch.ones(32).unsqueeze(0) * 8
g3 = torch.ones(32).unsqueeze(0) * 16
g4 = torch.ones(32).unsqueeze(0) * 24
```

```python
g1 = torch.Tensor(range(0, 32)).unsqueeze(0)
g2 = torch.Tensor(range(0, 32)).unsqueeze(0) 
g3 = torch.Tensor(range(0, 32)).unsqueeze(0)
g4 = torch.Tensor(range(0, 32)).unsqueeze(0)
```

```python
generators = [g1, g2, g3, g4]
```

```python
images = torch.Tensor(100, 0)

for gen in generators:
    patches = torch.Tensor(0, 32)

    for i in range(100):
        patches = torch.cat((patches, gen))

    images = torch.cat((images, patches), 1)
```

# **Configuration 1**

```python
im = images.reshape(100, 8, 8, 2)

fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].imshow(im[0,:,:,0], cmap="Blues", vmin=0, vmax=32)
axs[1].imshow(im[0,:,:,1], cmap="Blues", vmin=0, vmax=32)
plt.show()
```

```python
print(im[0,:,:,0])
print(im[0,:,:,1])
```
